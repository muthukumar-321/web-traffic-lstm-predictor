# Website Traffic Prediction
We will generate and analyze synthetic daily page view data for 5 articles over a 365-day period, then prepare a model to predict future traffic for efficient server resource allocation.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

np.random.seed(42)
dates = pd.date_range(start='2023-01-01', periods=365, freq='D')
articles = [f'Article_{i}' for i in range(1, 6)]

data = []
for article in articles:
    base = np.random.randint(500, 1500)
    trend = np.linspace(0, np.random.randint(100, 500), 365)
    weekly_pattern = 100 * np.sin(2 * np.pi * dates.dayofweek / 7)
    noise = np.random.normal(0, 50, 365)

    page_views = np.maximum(0, base + trend + weekly_pattern + noise).astype(int)

    for date, views in zip(dates, page_views):
        data.append({'Date': date, 'Article': article, 'PageViews': views})

df = pd.DataFrame(data)
display(df.head())
print(df.info())
```

```python
plt.figure(figsize=(14, 7))
sns.lineplot(data=df, x='Date', y='PageViews', hue='Article')
plt.title('Daily Page Views for 5 Articles over 365 Days')
plt.xlabel('Date')
plt.ylabel('Page Views')
plt.legend(title='Articles')
plt.grid(True)
plt.show()
```

## Feature Engineering
To forecast future page views, we will create lag features (page views from previous days) and rolling window statistics (e.g., 7-day moving average).

```python
df = df.sort_values(by=['Article', 'Date']).reset_index(drop=True)
for lag in [1, 2, 7]:
    df[f'Lag_{lag}'] = df.groupby('Article')['PageViews'].shift(lag)
df['Rolling_Mean_7'] = df.groupby('Article')['PageViews'].transform(lambda x: x.shift(1).rolling(7).mean())

df['DayOfWeek'] = df['Date'].dt.dayofweek
df['Month'] = df['Date'].dt.month
df_ml = df.dropna().reset_index(drop=True)
display(df_ml.head())
```

## Model Training and Evaluation
We will split our data chronologically into training and testing sets to evaluate our model's performance realistically. We will use a Random Forest Regressor to predict the page views.

```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
max_date = df_ml['Date'].max()
split_date = df_ml['Date'].min() + pd.Timedelta(days=300)

train_df = df_ml[df_ml['Date'] <= split_date]
test_df = df_ml[df_ml['Date'] > split_date]
features = ['Lag_1', 'Lag_2', 'Lag_7', 'Rolling_Mean_7', 'DayOfWeek', 'Month']
target = 'PageViews'

X_train, y_train = train_df[features], train_df[target]
X_test, y_test = test_df[features], test_df[target]
model = RandomForestRegressor(n_estimators=100, random_state=42)
model.fit(X_train, y_train)
y_pred = model.predict(X_test)
mae = mean_absolute_error(y_test, y_pred)
mse = mean_squared_error(y_test, y_pred)
rmse = np.sqrt(mse)
r2 = r2_score(y_test, y_pred)

print(f"Mean Absolute Error: {mae:.2f}")
print(f"Root Mean Squared Error: {rmse:.2f}")
print(f"R2 Score: {r2:.2f}")
```

```python
sample_article = 'Article_1'
sample_test = test_df[test_df['Article'] == sample_article].copy()
sample_test['Predicted'] = model.predict(sample_test[features])

plt.figure(figsize=(14, 6))
plt.plot(sample_test['Date'], sample_test['PageViews'], label='Actual', marker='o')
plt.plot(sample_test['Date'], sample_test['Predicted'], label='Predicted', marker='x', linestyle='--')
plt.title(f'Actual vs Predicted Page Views for {sample_article}')
plt.xlabel('Date')
plt.ylabel('Page Views')
plt.legend()
plt.grid(True)
plt.show()
```

## Sequential Model Setup and Chronological Sequence Generation
To capture temporal dependencies directly, we will structure our data into chronological sequences. We will build a sequential neural network (using TensorFlow/Keras) to forecast future traffic, addressing tasks 2, 3, 4, and 5.

```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense, Dropout
from sklearn.preprocessing import MinMaxScaler

sequence_length = 14

def create_sequences(data, seq_length):
    X, y = [], []
    for i in range(len(data) - seq_length):
        X.append(data[i:(i + seq_length)])
        y.append(data[i + seq_length])
    return np.array(X), np.array(y)


scalers = {}
article_data_train = {}
article_data_test = {}

target_article = 'Article_1'
art_df = df[df['Article'] == target_article].sort_values('Date').copy()
scaler = MinMaxScaler()
scaled_views = scaler.fit_transform(art_df[['PageViews']]).flatten()
train_size = int(len(scaled_views) * 0.8)
train_scaled = scaled_views[:train_size]
test_scaled = scaled_views[train_size:]

X_train_seq, y_train_seq = create_sequences(train_scaled, sequence_length)
X_test_seq, y_test_seq = create_sequences(test_scaled, sequence_length)
X_train_seq = np.reshape(X_train_seq, (X_train_seq.shape[0], X_train_seq.shape[1], 1))
X_test_seq = np.reshape(X_test_seq, (X_test_seq.shape[0], X_test_seq.shape[1], 1))

print(f"Train sequences shape: {X_train_seq.shape}")
print(f"Test sequences shape: {X_test_seq.shape}")
```

### Training the LSTM Sequential Model
We build an LSTM network. LSTMs are well-suited for this task as they have internal memory cells capable of capturing long-term dependencies and seasonal trends (like weekly patterns) across time steps.

```python
model_lstm = Sequential([
    LSTM(50, activation='relu', input_shape=(sequence_length, 1), return_sequences=False),
    Dropout(0.1),
    Dense(1)
])

model_lstm.compile(optimizer='adam', loss='mse')

history = model_lstm.fit(
    X_train_seq, y_train_seq,
    epochs=25,
    batch_size=16,
    validation_split=0.1,
    verbose=1
)
```

### Sequential Model Evaluation and Visualization on Unseen Dates

```python
predictions_scaled = model_lstm.predict(X_test_seq)

predictions = scaler.inverse_transform(predictions_scaled).flatten()
actuals = scaler.inverse_transform(y_test_seq.reshape(-1, 1)).flatten()

seq_mae = mean_absolute_error(actuals, predictions)
seq_rmse = np.sqrt(mean_squared_error(actuals, predictions))

print(f"Sequential Model (LSTM) - MAE: {seq_mae:.2f}")
print(f"Sequential Model (LSTM) - RMSE: {seq_rmse:.2f}")

test_dates = art_df['Date'].iloc[train_size + sequence_length:].reset_index(drop=True)
results_df = pd.DataFrame({
    'Date': test_dates,
    'Actual': actuals,
    'Predicted': predictions
})

plt.figure(figsize=(14, 6))
plt.plot(results_df['Date'], results_df['Actual'], label='Actual Traffic (Unseen Dates)', marker='o')
plt.plot(results_df['Date'], results_df['Predicted'], label='Predicted Traffic (LSTM)', marker='x', linestyle='--')
plt.title(f'LSTM Sequential Forecasting on Unseen Dates for {target_article}')
plt.xlabel('Date')
plt.ylabel('Page Views')
plt.legend()
plt.grid(True)
plt.show()
```
